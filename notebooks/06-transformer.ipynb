# Post-normalization transformer block

Self-attention and ReLU feed-forward sublayers each have residual connections and layer normalization. Compare whole block with PyTorch.

```python
import numpy as np
from transformer_scratch import *
from transformer_scratch.reference import ref_attention, ref_multi, ref_block
r=np.random.default_rng(11)
```

```python
m=TransformerBlock(8,2,16,r)
x=r.normal(size=(1,6,8))
a,w=m.forward(x)
np.testing.assert_allclose(a,ref_block(m,x),atol=1e-12)
print(a.shape)
```
