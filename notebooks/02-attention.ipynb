# Scaled dot-product attention

Scale QKᵀ by the square root of the head dimension; softmax across keys; multiply by V. Compare with PyTorch SDPA.

```python
import numpy as np
from transformer_scratch import *
from transformer_scratch.reference import ref_attention, ref_multi, ref_block
r=np.random.default_rng(11)
```

```python
q,k,v=r.normal(size=(3,1,2,4,3))
a,w=attention(q,k,v)
np.testing.assert_allclose(a,ref_attention(q,k,v),atol=1e-12)
print(w)
```
