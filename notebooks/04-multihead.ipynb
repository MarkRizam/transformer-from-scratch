# Multi-head self-attention

Independent projections are split into heads, attended, concatenated and projected. Match weights to PyTorch.

```python
import numpy as np
from transformer_scratch import *
from transformer_scratch.reference import ref_attention, ref_multi, ref_block
r=np.random.default_rng(11)
```

```python
x=r.normal(size=(2,5,8))
m=MultiHead(8,2,r)
a,w=m.forward(x)
b,z=ref_multi(m,x)
np.testing.assert_allclose(a,b,atol=1e-12)
np.testing.assert_allclose(w,z,atol=1e-12)
print(a.shape,w.shape)
```
