# Layer normalization

Normalize features of each token using population variance and learned affine parameters; do not normalize across tokens.

```python
import numpy as np
from transformer_scratch import *
from transformer_scratch.reference import ref_attention, ref_multi, ref_block
r=np.random.default_rng(11)
```

```python
x=r.normal(size=(2,4,8))
y=layer_norm(x,np.ones(8),np.zeros(8))
print(y.mean(-1),y.var(-1))
np.testing.assert_allclose(y.mean(-1),0,atol=1e-12)
```
