# Causal prefix invariance

Perturb future tokens; causal outputs at earlier positions must not change. This is a forward-pass property, not a language-generation evaluation.

```python
import numpy as np
from transformer_scratch import *
from transformer_scratch.reference import ref_attention, ref_multi, ref_block
r=np.random.default_rng(11)
```

```python
m=TransformerBlock(8,2,16,r)
x=r.normal(size=(1,6,8)); y=x.copy(); y[:,3:]+=5
a=m.forward(x,causal_mask(6))[0]; b=m.forward(y,causal_mask(6))[0]
np.testing.assert_allclose(a[:,:3],b[:,:3],atol=1e-12)
print(np.max(abs(a[:,:3]-b[:,:3])))
```
